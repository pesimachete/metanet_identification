```python

def compute_acf(x, max_lag=1000):
    """Computes the Autocorrelation Function using numpy."""
    x = np.asarray(x)
    n = len(x)
    mean = np.mean(x)
    var = np.var(x)
    if var == 0.0:
        return np.zeros(max_lag)

    x_centered = x - mean
    cov = np.correlate(x_centered, x_centered, mode="full")[n - 1 :]
    acf = cov[:max_lag] / (var * n)
    return acf

```

```python
filename = "../../mcmc_results_20blocks_20260616_153655.pkl"
with open(filename, "rb") as f:
    data = pickle.load(f)
burn_in = 1000
history = data["history"]
post_burn_in = history[burn_in:]
n_samples = len(post_burn_in)

print(compute_acf(post_burn_in, max_lag=100))

```
